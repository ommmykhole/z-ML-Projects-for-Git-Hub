```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.DataFrame({
    "City": ["Pune", "Mumbai", "Delhi", "Pune"]
})

encoder = OneHotEncoder()

encoded = encoder.fit_transform(df[["City"]])

encoded_df = pd.DataFrame(
    encoded.toarray(),
    columns=encoder.get_feature_names_out()
)

print(encoded_df)
```

```python
df = pd.DataFrame({
    "Food": ["Pizza", "Burger", "Pizza", "Sandwich"],
    "Drink": ["Tea", "Coffee", "Juice", "Tea"]
})

encoder = OneHotEncoder(sparse_output=False)

encoded = encoder.fit_transform(df)

encoded_df = pd.DataFrame(
    encoded,
    columns = encoder.get_feature_names_out()
)

print(encoded_df)
```

```python
df = pd.DataFrame({
    "Gender": ["Male", "Female", "Male", "Female"],
    "Department": ["HR", "IT", "Sales", "HR"],
    "City": ["Pune", "Mumbai", "Delhi", "Pune"],
    "Age": [21, 22, 20, 23],
    "Salary": [40000, 50000, 45000, 60000]
})

columns = ["Gender","Department","City"]
col1 = ["Age","Salary"]

encoded = OneHotEncoder()

encoder = encoded.fit_transform(df[columns])

encoded_df = pd.DataFrame(
    encoder.toarray(),
    columns = encoded.get_feature_names_out()
)

final_data = pd.concat([df[col1],encoded_df],axis=1)

print(final_data)
```

```python
df = pd.DataFrame({
    "Color": ["Red", "Blue", "Green", "Blue"],
    "Brand": ["Nike", "Puma", "Nike", "Adidas"],
    "City": ["Pune", "Mumbai", "Delhi", "Pune"],
    "Price": [5000, 3000, 4500, 6000],
    "Rating": [4.5, 4.2, 4.8, 4.1]
})

columns = ["Color","Brand","City"]
col1 = ["Price","Rating"]

encoder = OneHotEncoder(sparse_output=False)

encoded = encoder.fit_transform(df[columns])

encoded_df = pd.DataFrame(
    encoded,
    columns = encoder.get_feature_names_out()
)

DataSet = pd.concat([df[col1],encoded_df],axis=1)
print(DataSet)
```
