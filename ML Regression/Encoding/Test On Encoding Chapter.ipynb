```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
import numpy as pn
```

```python
df = pd.DataFrame({
    "Gender": ["Male", "Female", "Male", "Female"]
})

le = LabelEncoder()

df["Gender"] = le.fit_transform(df["Gender"])

print(df)
```

```python
df = pd.DataFrame({
    "City": ["Pune", "Mumbai", "Delhi", "Pune"]
})

encoder = OneHotEncoder()

encoded = encoder.fit_transform(df[["City"]])

encoded_data = pd.DataFrame(
    encoded.toarray(),
    columns = encoder.get_feature_names_out()
)

print(encoded_data)
```

```python
df = pd.DataFrame({
    "Gender": ["Male", "Female", "Male"],
    "Education": ["School", "College", "Master"],
    "City": ["Pune", "Delhi", "Mumbai"]
})

le = LabelEncoder()
df["Gender"] = le.fit_transform(df["Gender"])

encoder1 = OrdinalEncoder(
    categories = [["School","College","Master"]]
)

df[["Education"]] = encoder1.fit_transform(df[["Education"]])

encoder2 = OneHotEncoder()
encoded2 = encoder2.fit_transform(df[["City"]])

encoded_df = pd.DataFrame(
    encoded2.toarray(),
    columns = encoder2.get_feature_names_out()
)

print(df)
print(encoded_df)
```

```python
df = pd.DataFrame({
    "Gender": ["Male", "Female", "Male", "Female"],
    "City": ["Pune", "Delhi", "Mumbai", "Pune"],
    "Education": ["School", "College", "Master", "PhD"],
    "Priority": ["Low", "Medium", "High", "Low"],
    "Department": ["HR", "IT", "Sales", "HR"],
    "Salary": [40000, 50000, 60000, 45000],
    "Age": [21, 22, 23, 20]
})


le = LabelEncoder()
df["Gender"] = le.fit_transform(df["Gender"])


encoder1 = OneHotEncoder()
encoded1 = encoder1.fit_transform(df[["City","Department"]])
encoded_df = pd.DataFrame(
    encoded1.toarray(),
    columns = encoder1.get_feature_names_out(["City","Department"])
)


encoded2 = OrdinalEncoder(
    categories = [
        ["School","College","Master","PhD"],
        ["Low","Medium","High"]
    ]
)
df[["Education","Priority"]] = encoded2.fit_transform(df[["Education","Priority"]])



print(df)
print("="*40)
print(encoded_df)
```
