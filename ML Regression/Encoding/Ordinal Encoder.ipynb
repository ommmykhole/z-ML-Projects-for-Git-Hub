```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    "Size": ["M", "L", "S", "XL", "M"]
})

encoder = OrdinalEncoder(
    categories=[["S", "M", "L", "XL"]]
)

df["Size"] = encoder.fit_transform(df[["Size"]])

print(df)
```

```python
df = pd.DataFrame({
    "Performance": ["Poor", "Average", "Good", "Excellent", "Good"]
})

encoder = OrdinalEncoder(
    categories=[["Poor","Average","Good","Excellent"]]
)

df["Performance"] = encoder.fit_transform(df[["Performance"]])

print(df)
```

```python
df = pd.DataFrame({
    "Education": [
        "School",
        "College",
        "Master",
        "PhD"
    ],
    "Size": [
        "M",
        "XL",
        "S",
        "L"
    ],
    "Performance": [
        "Good",
        "Poor",
        "Excellent",
        "Average"
    ],
    "Age": [
        20,
        22,
        25,
        24
    ]
})

encoder = OrdinalEncoder(
    categories = [
        ["School","College","Master","PhD"],
        ["S","M","L","XL"],
        ["Poor","Average","Good","Excellent"]
    ]
)

df[["Education","Size","Performance"]] = encoder.fit_transform(df[["Education","Size","Performance"]])

print(df)
```
