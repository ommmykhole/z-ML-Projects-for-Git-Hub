```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    "Gender": ["Male", "Female", "Male"],
    "Result": ["Pass", "Fail", "Pass"]
})

le = LabelEncoder()

df["Gender"] = le.fit_transform(df["Gender"])

df["Result"] = le.fit_transform(df["Result"])

print(df)
```

```python
df = pd.DataFrame({
    "Country": ["India", "USA", "Japan", "India", "USA"],
    "Department": ["HR", "IT", "Sales", "HR", "IT"],
    "Payment": ["Cash", "UPI", "Card", "Cash", "UPI"],
    "Salary": [40000, 65000, 45000, 42000, 70000]
})

le = LabelEncoder()

df["Country"] = le.fit_transform(df["Country"])
df["Department"] = le.fit_transform(df["Department"])
df["Payment"] = le.fit_transform(df["Payment"])

print(df)
```

```python
df = pd.DataFrame({
    "Gender": ["Male", "Female", "Male", "Female"],
    "Department": ["HR", "IT", "Sales", "HR"],
    "Result": ["Pass", "Fail", "Pass", "Pass"],
    "Country": ["India", "USA", "Japan", "India"],
    "Age": [21, 22, 20, 23],
    "Salary": [40000, 60000, 50000, 45000]
})

le = LabelEncoder()

columns = ["Gender","Department","Result","Country"]

for col in columns:
    df[col] = le.fit_transform(df[col])

print(df)
```
