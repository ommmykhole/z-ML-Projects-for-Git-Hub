```python
import pandas as pd
from sklearn.feature_selection import mutual_info_classif

df = pd.DataFrame({
    "Age": [18, 20, 45, 50, 22, 48],
    "Salary": [25000, 30000, 80000, 90000, 28000, 85000],
    "Gender": [0, 1, 0, 1, 0, 1],
    "Purchased": [0, 0, 1, 1, 0, 1]
})

X = df.drop("Purchased", axis=1)
y = df["Purchased"]

mi_score = mutual_info_classif(X, y)

mi_score = pd.Series(mi_score, index=X.columns)

mi_score = mi_score.sort_values(ascending=False)

print(mi_score)
```

```python
mi_score = pd.DataFrame({
    "Features": X.columns,
    "MI Score": mi_score
})
```
