```python
# Step 1 : Import Libraries

import pandas as pd
from sklearn.feature_selection import VarianceThreshold


# Step 2 : Create Dataset

df = pd.DataFrame({
    "Gender": [1, 1, 1, 1, 1],
    "Age": [18, 20, 22, 24, 26],
    "Salary": [30000, 35000, 40000, 45000, 50000]
})


# Step 3 : Separate Features (X)

X = df


# Step 4 : Create VarianceThreshold Object

selector = VarianceThreshold(threshold=0)


# Step 5 : Fit the Model

selector.fit(X)


# Step 6 : Check Selected Features (True = Keep, False = Remove)

print(selector.get_support())


# Step 7 : Display Selected Column Names

selected_columns = X.columns[selector.get_support()]
print(selected_columns)


# Step 8 : Create Final Dataset

final_dataset = X[selected_columns]

print(final_dataset)
```

```python
# Step 1 : Import Library
from sklearn.feature_selection import VarianceThreshold

# Step 2 : Import Pandas
import pandas as pd

# Step 3 : Create Sample Dataset
df = pd.DataFrame({
    "Age": [20, 21, 22, 23, 24],
    "Salary": [30000, 32000, 31000, 33000, 34000],
    "Country": [1, 1, 1, 1, 1],
    "Gender": [0, 0, 0, 0, 1]
})

print("Original Dataset")
print(df)

# Step 4 : Create Variance Threshold Object
vt = VarianceThreshold(threshold=0.01)

# Step 5 : Fit and Transform the Dataset
X_new = vt.fit_transform(df)

# Step 6 : Get Selected Column Names
selected_columns = df.columns[vt.get_support()]

# Step 7 : Convert Back to DataFrame
new_df = pd.DataFrame(X_new, columns=selected_columns)

print("\nDataset After Variance Threshold")
print(new_df)
```
