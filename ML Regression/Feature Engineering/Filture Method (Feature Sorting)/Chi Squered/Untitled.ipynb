```python
from sklearn.feature_selection import chi2
import pandas as pd

df = pd.DataFrame({
    "Gender": [1, 0, 1, 0, 1, 0],
    "Education": [2, 1, 2, 1, 2, 1],
    "City": [1, 2, 1, 2, 1, 2],
    "Purchased": [1, 0, 1, 0, 1, 0]
})
X = df.drop("Purchased", axis=1)
y = df["Purchased"]

chi_score, p_value = chi2(X, y)
chi_df = pd.DataFrame({
    "Feature": X.columns,
    "Chi Score": chi_score,
    "P Value": p_value
})

chi_df = chi_df.sort_values(
    by="Chi Score",
    ascending=False
)

chi_df
```

```python
import pandas as pd

from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import chi2

df = pd.read_csv("loan.csv")

X = df.drop(columns=["Loan"])
y = df["Loan"]

selector = SelectKBest(score_func=chi2, k=3)

X_new = selector.fit_transform(X, y)

score = pd.DataFrame({
    "Feature": X.columns,
    "Score": selector.scores_
})

print(score)
```
