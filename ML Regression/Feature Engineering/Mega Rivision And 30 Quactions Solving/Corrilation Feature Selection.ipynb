```python
# Create Correlation Matrix
matrix = X.corr(numeric_only=True)

# Display Heatmap (Optional)
sns.heatmap(matrix, annot=True, cmap="coolwarm")
plt.show()

# Keep only Upper Triangle
upper_triangle = matrix.where(
    np.triu(np.ones(matrix.shape), k=1).astype(bool)
)

# Find Highly Correlated Features
columns_to_drop = [
    column
    for column in upper_triangle.columns
    if any(abs(upper_triangle[column]) > 0.90)
]

print("Columns to Drop:")
print(columns_to_drop)

# Remove Highly Correlated Features
X_corr = X.drop(columns=columns_to_drop)

print("\nRemaining Features")
print(X_corr.columns)
```
