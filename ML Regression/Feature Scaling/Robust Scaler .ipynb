```python
from sklearn.preprocessing import RobustScaler
import pandas as pd

df = pd.DataFrame({
    "Age": [20, 22, 24, 25, 500]
})

scaler = RobustScaler()

df["Age"] = scaler.fit_transform(df[["Age"]])

print(df)
```
