```python
import pandas as pd
from sklearn.preprocessing import Normalizer

df = pd.DataFrame({
    "Maths": [3, 6],
    "Science": [4, 8]
})

print("Original Data:")
print(df)

scaler = Normalizer()

scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(scaled_data, columns=df.columns)

print("\nNormalized Data:")
print(scaled_df)
```
