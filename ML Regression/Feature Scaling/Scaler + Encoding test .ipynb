```python
import pandas as pd

df = pd.DataFrame({
    "Gender": ["Male", "Female", "Male", "Female"],
    "Education": ["School", "College", "Masters", "PhD"],
    "Salary": [25000, 40000, 55000, 800000],
    "Age": [20, 25, 30, 35]
})


le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])


coder = OrdinalEncoder(
    categories=[["School","College","Masters","PhD"]]
)

df[["Education"]] = coder.fit_transform(df[["Education"]])


scaler = RobustScaler()
scaled = scaler.fit_transform(df[["Salary"]])


scaler1 = MinMaxScaler()
scaled1 = scaler1.fit_transform(df[["Age"]])

scaled1 = pd.DataFrame(scaled1, columns=["Scaled_1"])
scaled = pd.DataFrame(scaled, columns=["Scaled_2"])

final_dataset = pd.concat([df, scaled1, scaled], axis=1)

print(final_dataset)
```
