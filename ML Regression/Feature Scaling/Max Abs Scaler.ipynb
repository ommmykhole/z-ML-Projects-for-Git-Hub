```python
from sklearn.preprocessing import MaxAbsScaler
import pandas as pd

df = pd.DataFrame({
    "Marks": [-100, -50, 0, 50, 100]
})

scaler = MaxAbsScaler()

df["Marks"] = scaler.fit_transform(df[["Marks"]])

print(df)
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, MaxAbsScaler

df = pd.DataFrame({
    "Feature": [0, 0, 10, 0, 20, 0, 30]
})

# MinMaxScaler
scaled1 = MinMaxScaler()
scaler1 = scaled1.fit_transform(df)

# StandardScaler
scaled2 = StandardScaler()
scaler2 = scaled2.fit_transform(df)

# MaxAbsScaler
scaled3 = MaxAbsScaler()
scaler3 = scaled3.fit_transform(df)

print("MinMaxScaler:")
print(scaler1)

print("\nStandardScaler:")
print(scaler2)

print("\nMaxAbsScaler:")
print(scaler3)
```
