```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

data = pd.DataFrame({
    "Age": [20, 30, 40, 50, 60]
})

scaler = MinMaxScaler()

scaled = scaler.fit_transform(data)

print(scaled)
```
