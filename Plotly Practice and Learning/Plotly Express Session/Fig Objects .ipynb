```python
import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px

# =========================================================
# PAGE CONFIG
# =========================================================

st.set_page_config(
    page_title="Sales Analytics Dashboard",
    page_icon="📊",
    layout="wide"
)


# =========================================================
# CUSTOM CSS
# =========================================================

st.markdown("""
<style>

.stApp {
    background: #0B1120;
    color: #E5E7EB;
}

.block-container {
    padding-top: 2rem;
    padding-bottom: 2rem;
    max-width: 1500px;
}

.dashboard-title {
    font-size: 38px;
    font-weight: 800;
    color: #F8FAFC;
    margin-bottom: 5px;
}

.dashboard-subtitle {
    color: #94A3B8;
    font-size: 16px;
    margin-bottom: 25px;
}

.kpi-card {
    background: #111827;
    padding: 22px;
    border-radius: 15px;
    border: 1px solid #1F2937;
    box-shadow: 0px 6px 20px rgba(0,0,0,0.25);
}

.kpi-title {
    color: #94A3B8;
    font-size: 14px;
}

.kpi-value {
    color: #F8FAFC;
    font-size: 28px;
    font-weight: 800;
    margin-top: 5px;
}

</style>
""", unsafe_allow_html=True)


# =========================================================
# SAMPLE DATA
# =========================================================

np.random.seed(42)

months = [
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug",
    "Sep", "Oct", "Nov", "Dec"
]

regions = ["West", "North", "South", "East"]
segments = ["Premium", "Regular", "Budget"]

rows = []

for region in regions:
    for segment in segments:
        for month in months:

            sales = np.random.randint(50000, 150000)
            profit = sales * np.random.uniform(0.12, 0.30)
            customers = np.random.randint(100, 600)

            rows.append({
                "Month": month,
                "Region": region,
                "Segment": segment,
                "Sales": sales,
                "Profit": profit,
                "Customers": customers
            })

df = pd.DataFrame(rows)


# =========================================================
# HEADER
# =========================================================

st.markdown(
    '<div class="dashboard-title">Sales & Customer Analytics</div>',
    unsafe_allow_html=True
)

st.markdown(
    '<div class="dashboard-subtitle">'
    'Interactive business performance dashboard'
    '</div>',
    unsafe_allow_html=True
)


# =========================================================
# SLICERS
# =========================================================

col1, col2, col3 = st.columns(3)

with col1:

    selected_region = st.selectbox(
        "🌍 Region",
        ["All"] + regions
    )

with col2:

    selected_segment = st.selectbox(
        "👥 Customer Segment",
        ["All"] + segments
    )

with col3:

    selected_month = st.selectbox(
        "📅 Month",
        ["All"] + months
    )


# =========================================================
# FILTER DATA
# =========================================================

filtered_df = df.copy()

if selected_region != "All":
    filtered_df = filtered_df[
        filtered_df["Region"] == selected_region
    ]

if selected_segment != "All":
    filtered_df = filtered_df[
        filtered_df["Segment"] == selected_segment
    ]

if selected_month != "All":
    filtered_df = filtered_df[
        filtered_df["Month"] == selected_month
    ]


# =========================================================
# KPI CALCULATIONS
# =========================================================

total_sales = filtered_df["Sales"].sum()

total_profit = filtered_df["Profit"].sum()

total_customers = filtered_df["Customers"].sum()

avg_sales = filtered_df["Sales"].mean()


# =========================================================
# KPI CARDS
# =========================================================

k1, k2, k3, k4 = st.columns(4)

with k1:

    st.markdown(f"""
    <div class="kpi-card">
        <div class="kpi-title">TOTAL REVENUE</div>
        <div class="kpi-value">₹{total_sales:,.0f}</div>
    </div>
    """, unsafe_allow_html=True)

with k2:

    st.markdown(f"""
    <div class="kpi-card">
        <div class="kpi-title">TOTAL PROFIT</div>
        <div class="kpi-value">₹{total_profit:,.0f}</div>
    </div>
    """, unsafe_allow_html=True)

with k3:

    st.markdown(f"""
    <div class="kpi-card">
        <div class="kpi-title">CUSTOMERS</div>
        <div class="kpi-value">{total_customers:,}</div>
    </div>
    """, unsafe_allow_html=True)

with k4:

    st.markdown(f"""
    <div class="kpi-card">
        <div class="kpi-title">AVERAGE SALES</div>
        <div class="kpi-value">₹{avg_sales:,.0f}</div>
    </div>
    """, unsafe_allow_html=True)


st.markdown("<br>", unsafe_allow_html=True)


# =========================================================
# CHART 1 — MONTHLY SALES
# =========================================================

monthly_sales = (
    filtered_df
    .groupby("Month", as_index=False)["Sales"]
    .sum()
)

fig1 = px.line(
    monthly_sales,
    x="Month",
    y="Sales",
    markers=True,
    title="Monthly Revenue Trend"
)

fig1.update_traces(line_width=3)

fig1.update_layout(
    template="plotly_dark",
    paper_bgcolor="#111827",
    plot_bgcolor="#111827",
    font=dict(color="#E5E7EB"),
    title_font_size=20
)


# =========================================================
# CHART 2 — REGION SALES
# =========================================================

region_sales = (
    filtered_df
    .groupby("Region", as_index=False)["Sales"]
    .sum()
)

fig2 = px.bar(
    region_sales,
    x="Region",
    y="Sales",
    title="Revenue by Region"
)

fig2.update_layout(
    template="plotly_dark",
    paper_bgcolor="#111827",
    plot_bgcolor="#111827",
    font=dict(color="#E5E7EB"),
    title_font_size=20
)


# =========================================================
# CHART 3 — CUSTOMER SEGMENT
# =========================================================

segment_sales = (
    filtered_df
    .groupby("Segment", as_index=False)["Sales"]
    .sum()
)

fig3 = px.pie(
    segment_sales,
    names="Segment",
    values="Sales",
    hole=0.55,
    title="Revenue by Customer Segment"
)

fig3.update_layout(
    template="plotly_dark",
    paper_bgcolor="#111827",
    font=dict(color="#E5E7EB"),
    title_font_size=20
)


# =========================================================
# CHART 4 — PROFIT TREND
# =========================================================

monthly_profit = (
    filtered_df
    .groupby("Month", as_index=False)["Profit"]
    .sum()
)

fig4 = px.line(
    monthly_profit,
    x="Month",
    y="Profit",
    markers=True,
    title="Monthly Profit Trend"
)

fig4.update_traces(line_width=3)

fig4.update_layout(
    template="plotly_dark",
    paper_bgcolor="#111827",
    plot_bgcolor="#111827",
    font=dict(color="#E5E7EB"),
    title_font_size=20
)


# =========================================================
# CHART 5 — SALES VS PROFIT
# =========================================================

fig5 = px.scatter(
    filtered_df,
    x="Sales",
    y="Profit",
    size="Customers",
    hover_data=["Region", "Segment", "Month"],
    title="Sales vs Profit"
)

fig5.update_layout(
    template="plotly_dark",
    paper_bgcolor="#111827",
    plot_bgcolor="#111827",
    font=dict(color="#E5E7EB"),
    title_font_size=20
)


# =========================================================
# CHART 6 — SEGMENT SALES
# =========================================================

segment_sales2 = (
    filtered_df
    .groupby("Segment", as_index=False)["Sales"]
    .sum()
    .sort_values("Sales", ascending=False)
)

fig6 = px.bar(
    segment_sales2,
    x="Sales",
    y="Segment",
    orientation="h",
    title="Top Customer Segments"
)

fig6.update_layout(
    template="plotly_dark",
    paper_bgcolor="#111827",
    plot_bgcolor="#111827",
    font=dict(color="#E5E7EB"),
    title_font_size=20
)


# =========================================================
# DISPLAY 6 CHARTS
# =========================================================

c1, c2 = st.columns(2)

with c1:
    st.plotly_chart(
        fig1,
        use_container_width=True
    )

with c2:
    st.plotly_chart(
        fig2,
        use_container_width=True
    )


c3, c4 = st.columns(2)

with c3:
    st.plotly_chart(
        fig3,
        use_container_width=True
    )

with c4:
    st.plotly_chart(
        fig4,
        use_container_width=True
    )


c5, c6 = st.columns(2)

with c5:
    st.plotly_chart(
        fig5,
        use_container_width=True
    )

with c6:
    st.plotly_chart(
        fig6,
        use_container_width=True
    )
```
