```python
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots


# =========================
# KPI VALUES
# =========================

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_units = df["Units_Sold"].sum()


# =========================
# CREATE DASHBOARD
# =========================

fig = make_subplots(
    rows=3,
    cols=2,
    specs=[
        [
            {"type": "indicator"},
            {"type": "indicator"}
        ],
        [
            {"type": "indicator"},
            {"type": "pie"}
        ],
        [
            {"type": "bar"},
            {"type": "scatter"}
        ]
    ],
    subplot_titles=(
        "",
        "",
        "",
        "Region Wise Sales",
        "Product Wise Sales",
        "Sales vs Profit"
    )
)


# =========================
# KPI 1 — SALES
# =========================

fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_sales,
        title={"text": "Total Sales"}
    ),
    row=1,
    col=1
)


# =========================
# KPI 2 — PROFIT
# =========================

fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_profit,
        title={"text": "Total Profit"}
    ),
    row=1,
    col=2
)


# =========================
# KPI 3 — UNITS SOLD
# =========================

fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_units,
        title={"text": "Total Units Sold"}
    ),
    row=2,
    col=1
)


# =========================
# PIE — REGION SALES
# =========================

region_sales = df.groupby("Region", as_index=False)["Sales"].sum()

fig.add_trace(
    go.Pie(
        labels=region_sales["Region"],
        values=region_sales["Sales"],
        hole=0.4
    ),
    row=2,
    col=2
)


# =========================
# BAR — PRODUCT SALES
# =========================

product_sales = df.groupby("Product", as_index=False)["Sales"].sum()

fig.add_trace(
    go.Bar(
        x=product_sales["Product"],
        y=product_sales["Sales"],
        name="Sales"
    ),
    row=3,
    col=1
)


# =========================
# SCATTER — SALES VS PROFIT
# =========================

fig.add_trace(
    go.Scatter(
        x=df["Sales"],
        y=df["Profit"],
        mode="markers",
        text=df["Product"],
        name="Products"
    ),
    row=3,
    col=2
)


# =========================
# DASHBOARD LAYOUT
# =========================

fig.update_layout(
    title={
        "text": "📊 Sales Performance Dashboard",
        "x": 0.5
    },
    height=1000,
    width=1200,
    template="plotly_white",
    showlegend=False
)


fig.show()
```
