```python
import pandas as pd
import plotly.graph_objects as go

from plotly.subplots import make_subplots


# ============================================================
# DATA
# ============================================================

df_sales = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120000, 150000, 140000, 240000, 180000, 210000]
})


df_product = pd.DataFrame({
    "Product": ["Shirts", "Pants", "Jeans", "Shoes", "Jackets"],
    "Sales": [85000, 72000, 95000, 60000, 45000]
})


df_category = pd.DataFrame({
    "Category": ["Men", "Women", "Kids", "Accessories"],
    "Sales": [180000, 150000, 90000, 60000]
})


df_units = pd.DataFrame({
    "Product": ["Shirts", "Pants", "Jeans", "Shoes", "Jackets"],
    "Units": [420, 350, 390, 280, 190]
})


# ============================================================
# CREATE SUBPLOTS
# ============================================================

fig = make_subplots(

    rows=2,
    cols=2,

    subplot_titles=(
        "Monthly Sales",
        "Product-wise Sales",
        "Sales Distribution",
        "Units Sold"
    ),

    specs=[
        [{"type": "xy"}, {"type": "xy"}],
        [{"type": "domain"}, {"type": "xy"}]
    ]
)


# ============================================================
# 1️⃣ MONTHLY SALES - LINE CHART
# ============================================================

fig.add_trace(

    go.Scatter(
        x=df_sales["Month"],
        y=df_sales["Sales"],
        mode="lines+markers",
        name="Monthly Sales"
    ),

    row=1,
    col=1
)


# ============================================================
# 2️⃣ PRODUCT SALES - BAR CHART
# ============================================================

fig.add_trace(

    go.Bar(
        x=df_product["Product"],
        y=df_product["Sales"],
        name="Product Sales"
    ),

    row=1,
    col=2
)


# ============================================================
# 3️⃣ SALES DISTRIBUTION - DONUT CHART
# ============================================================

fig.add_trace(

    go.Pie(
        labels=df_category["Category"],
        values=df_category["Sales"],
        hole=0.5,
        name="Sales Distribution"
    ),

    row=2,
    col=1
)


# ============================================================
# 4️⃣ UNITS SOLD - BAR CHART
# ============================================================

fig.add_trace(

    go.Bar(
        x=df_units["Product"],
        y=df_units["Units"],
        name="Units Sold"
    ),

    row=2,
    col=2
)


# ============================================================
# UPDATE TRACES
# ============================================================

fig.update_traces(

    marker=dict(
        line=dict(
            width=1
        )
    )
)


# ============================================================
# X-AXIS
# ============================================================

fig.update_xaxes(

    title="Month",
    showgrid=True,

    row=1,
    col=1
)


fig.update_xaxes(

    title="Product",
    showgrid=False,

    tickangle=-30,

    row=1,
    col=2
)


fig.update_xaxes(

    title="Product",
    showgrid=False,

    tickangle=-30,

    row=2,
    col=2
)


# ============================================================
# Y-AXIS
# ============================================================

fig.update_yaxes(

    title="Sales",
    showgrid=True,

    tickformat="₹,.0f",

    row=1,
    col=1
)


fig.update_yaxes(

    title="Sales",
    showgrid=False,

    tickformat="₹,.0f",

    row=1,
    col=2
)


fig.update_yaxes(

    title="Units",
    showgrid=False,

    row=2,
    col=2
)


# ============================================================
# ANNOTATION
# ============================================================

fig.add_annotation(

    x="Apr",
    y=240000,

    text="Highest Monthly Sales",

    showarrow=True,

    arrowhead=2,

    row=1,
    col=1
)


# ============================================================
# OVERALL LAYOUT
# ============================================================

fig.update_layout(

    title="Retail Business Dashboard",

    width=1100,
    height=800,

    paper_bgcolor="#111827",

    plot_bgcolor="#111827",

    font=dict(
        family="Arial",
        size=13,
        color="#F8FAFC"
    ),

    margin=dict(
        l=50,
        r=50,
        t=90,
        b=50
    ),

    showlegend=False
    
)


fig.show()
```
