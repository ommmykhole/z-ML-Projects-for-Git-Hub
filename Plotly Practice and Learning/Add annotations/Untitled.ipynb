```python
import pandas as pd
import plotly.graph_objects as go


df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120000, 150000, 140000, 240000, 180000, 210000]
})


fig = go.Figure()


# ---------------- TRACE ----------------

fig.add_trace(
    go.Scatter(
        x=df["Month"],
        y=df["Sales"],
        mode="lines+markers",
        name="Monthly Sales"
    )
)


# ---------------- LAYOUT ----------------

fig.update_layout(
    title="Monthly Sales Performance",
    width=800,
    height=600,

    margin=dict(
        l=40,
        r=40,
        t=60,
        b=40
    ),

    paper_bgcolor="#111827",
    plot_bgcolor="#1F2937",

    font=dict(
        family="Arial",
        size=13,
        color="#F8FAFC"
    )
)


# ---------------- X AXIS ----------------

fig.update_xaxes(
    title="Month",
    showline=True,
    showgrid=False,
    tickfont=dict(size=12),
    title_font=dict(size=15),

    categoryorder="array",
    categoryarray=["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
)


# ---------------- Y AXIS ----------------

fig.update_yaxes(
    title="Sales",
    showline=True,
    showgrid=False,
    zeroline=True,

    tickfont=dict(size=12),
    title_font=dict(size=15),

    tickformat="₹,.0f",
    range=[0, 300000],
    dtick=50000
)


# ---------------- ANNOTATION ----------------

# Annotation ka use chart par additional information / insight
# dikhane ke liye kiya jata hai.

fig.add_annotation(
    x="Apr",
    y=240000,
    text="Highest Sales",
    showarrow=True,

    # Arrow ko customize karne ke liye
    arrowhead=2,

    # Annotation text ki position ko adjust karne ke liye
    ax=40,
    ay=-50,

    font=dict(
        size=14
    )
)


fig.show()
```
