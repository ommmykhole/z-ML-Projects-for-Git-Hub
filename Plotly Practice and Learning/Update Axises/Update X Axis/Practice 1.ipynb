```python
import pandas as pd
import plotly.graph_objects as go


df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [120, 180, 150, 240, 300]
})


fig = go.Figure()


fig.add_trace(
    go.Bar(
        x=df["Month"],
        y=df["Sales"],
        name="Monthly Sales"
    )
)

fig.update_layout(
    title="Month Wise Sales",
    width=600,
    height=670,
    margin=dict(l=40,r=40,t=40,b=30),
    paper_bgcolor="#1E293B",   # Main outer background
    plot_bgcolor="#1E293B",    # Chart/Paper area
    font=dict(
        color="#F8FAFC"
    ),
    template="plotly_dark"
)

fig.update_traces(
    opacity=0.8,
    marker_line_width=1,
    hovertemplate="%{x}: %{y}<extra></extra>"
)

fig.update_xaxes(
    title="Month",
    tickangle=45,
    showgrid=False,
    showline=True,
    zeroline=True,
    tickfont=dict(size=12),
    title_font=dict(size=14),
    categoryorder="array",
    categoryarray=["Jan", "Feb", "Mar", "Apr", "May"]
)



fig.show()
```

```python
df1 = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 180, 150, 240, 210, 300]
})

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df1["Month"],
        y=df1["Sales"]
    )
)

fig.update_layout(
    title="Month Wise Sales",
    width=700,
    height=650,
    margin=dict(l=40,r=40,t=40,b=30),
    paper_bgcolor="#0F172A",   # Main outer background
    plot_bgcolor="#1E293B",    # Chart/Paper area
    font=dict(
        color="#F8FAFC"
    ),
    template="plotly_dark"
    
    
)

fig.update_traces(
    opacity=0.4,
    marker_line_width=1,
    
)

fig.update_xaxes(
    title="Month",
    tickangle=-35,
    showgrid=True,
    showline=True,
    tickfont=dict(size=15),
    title_font=dict(size=19),
    categoryorder="array",
    categoryarray=["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
    
)

fig.show()
```

```python
df2 = pd.DataFrame({
    "Product": ["Laptop", "Smartphone", "Headphones", "Keyboard", "Monitor"],
    "Units_Sold": [45, 80, 120, 65, 50]
})

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=df2["Product"],
        y=df2["Units_Sold"]
    )
)

fig.update_layout(
    title="Product Sales",
    width=700,
    height=670,
    margin=dict(l=40,r=40,t=40,b=30),
    paper_bgcolor="#0F172A",   # Main outer background
    plot_bgcolor="#1E293B",    # Chart/Paper area
    font=dict(
        color="#F8FAFC"
    ),
    template="plotly_dark"
)

fig.update_traces(
    marker=dict(
        color="#10B981",          # Bar color
        line=dict(
            color="#10B981",      # Bar border
            width=1
        )
    ),
    opacity=0.8,

    texttemplate="%{y}",
    textposition="outside"

   
)

fig.update_xaxes(
    title="Product",
    tickangle=0,
    showgrid=False,
    showline=True,
    tickfont=dict(size=12),
    title_font=dict(size=13)
)

fig.show()
```

```python
df3 = pd.DataFrame({
    "Department": ["Sales", "Marketing", "Finance", "HR", "IT"],
    "Revenue": [450000, 320000, 280000, 190000, 510000]
})

fig = go.Figure()

fig.add_trace(
    go.Pie(
        labels=df3["Department"],
        values=df3["Revenue"]
    )
)

fig.update_layout(
    title="Departments and Their Revenue",
    width=650,
    height=650,
    margin=dict(l=40,r=40,t=40,b=30),
    paper_bgcolor="#0F172A",   # Main outer background
    plot_bgcolor="#1E293B",    # Chart/Paper area
    font=dict(
        color="#F8FAFC"
    ),
    template="plotly_dark"
    
)

fig.update_traces(
    marker=dict(
        colors=["#2563EB", "#10B981", "#F59E0B", "#EF4444"],
        
    ),

    textinfo="label+percent",
    textposition="outside",

    pull=[0.05, 0, 0, 0],

    hovertemplate=
        "<b>%{label}</b><br>"
        "Value: %{value}<br>"
        "Share: %{percent}<extra></extra>"
)

fig.update_xaxes(
    title="Pie Chart",
    tickangle=0,
    showgrid=False,
    showline=False,
    tickfont=dict(size=14),
    title_font=dict(size=15)
)

fig.show()
```

```python
df4 = pd.DataFrame({
    "Store": ["Pune", "Mumbai", "Nashik", "Nagpur", "Kolhapur"],
    "Customers": [420, 680, 310, 450, 290]
})

fig = go.Figure()

fig.add_trace(
    go.Funnel(
        x=df4["Customers"],
        y=df4["Store"]
    )
)

fig.update_layout(
    title="Stores And Customers",
    width=700,
    height=650,
    margin=dict(l=30,r=30,t=40,b=40),
    paper_bgcolor="#0F172A",   # Main outer background
    plot_bgcolor="#1E293B",    # Chart/Paper area
    font=dict(
        color="#F8FAFC"
    ),
    template="plotly_dark"
    
)

fig.update_traces(
    marker=dict(
        color=["#2563EB", "#10B981", "#F59E0B", "#EF4444"],
        
    ),

    textinfo="label+value+percent initial",

    textposition="inside",

    hovertemplate=
        "<b>%{y}</b><br>"
        "Count: %{x}<br>"
        "Share: %{percentInitial}<extra></extra>"
)

fig.update_xaxes(
    title="Store Info",
    
)

fig.show()
```

```python
df5 = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Profit": [25000, 32000, 28000, 41000, 38000, 52000]
})

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df5["Month"],
        y=df5["Profit"]
    )
)

fig.update_layout(
    title="Month Wise Profit",
    width=700,
    height=650,
    margin=dict(l=30,r=30,t=40,b=40),
    paper_bgcolor="#0F172A",   # Main outer background
    plot_bgcolor="#1E293B",    # Chart/Paper area
    font=dict(
        color="#F8FAFC"
    ),
    template="plotly_dark"
    
)


fig.update_traces(
    line=dict(
        color="#2563EB",
        width=3,
        shape="spline"
    ),

    mode="lines+markers",

    marker=dict(
        size=8,
        color="#2563EB",
        line=dict(
            color="white",
            width=2
        )
    ),
    

    opacity=0.95,

    hovertemplate=
        "<b>%{x}</b><br>"
        "Profit: %{y}<extra></extra>"
)

fig.update_xaxes(
    title="Month",
    showline=True,
    tickfont=dict(size=12),
    title_font=dict(size=15),
    categoryorder="array",
    categoryarray=["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
)

fig.show()
```
