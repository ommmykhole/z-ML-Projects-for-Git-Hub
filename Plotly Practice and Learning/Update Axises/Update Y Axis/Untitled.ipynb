```python
import pandas as pd
import plotly.graph_objects as go


# --------------------------------
# 1. Data
# --------------------------------
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Revenue": [25000, 32000, 28000, 41000, 38000, 52000]
})


# --------------------------------
# 2. Create Figure
# --------------------------------
fig = go.Figure()


# --------------------------------
# 3. Add Bar Trace
# --------------------------------
fig.add_trace(
    go.Bar(
        x=df["Month"],
        y=df["Revenue"],
        name="Revenue"
    )
)


# --------------------------------
# 4. Update Traces
# --------------------------------
fig.update_traces(
    marker=dict(
        line=dict(
            color="white",
            width=1
        )
    ),
    opacity=0.9,

    hovertemplate=
        "<b>%{x}</b><br>"
        "Revenue: ₹%{y:,.0f}"
        "<extra></extra>"
)


# --------------------------------
# 5. Update Layout
# --------------------------------
fig.update_layout(
    title="Monthly Revenue",
    width=800,
    height=600,

    margin=dict(
        l=50,
        r=30,
        t=60,
        b=50
    ),

    paper_bgcolor="#0F172A",
    plot_bgcolor="#1E293B",

    font=dict(
        color="#F8FAFC"
    ),

    template="plotly_dark"
)


# --------------------------------
# 6. Update X-Axis
# --------------------------------
fig.update_xaxes(
    title="Month",

    tickangle=0,

    showgrid=False,

    showline=True,

    zeroline=True,

    tickfont=dict(
        size=12
    ),

    title_font=dict(
        size=15
    ),

    categoryorder="array",

    categoryarray=[
        "Jan",
        "Feb",
        "Mar",
        "Apr",
        "May",
        "Jun"
    ]
)


# --------------------------------
# 7. Update Y-Axis
# --------------------------------
fig.update_yaxes(
    title="Revenue (₹)",

    showgrid=True,

    showline=True,

    zeroline=True,

    tickfont=dict(
        size=12
    ),

    title_font=dict(
        size=15
    ),

    range=[0, 60000],

    dtick=10000,

    tickformat=",.0f"
)


# --------------------------------
# 8. Show Figure
# --------------------------------
fig.show()
```

```python
df1 = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Revenue": [25000, 32000, 28000, 41000, 38000, 52000]
})


fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df1["Month"],
        y=df1["Revenue"]
    )
)

fig.update_layout(
    title="Month Wise Revenue",
    width=700,
    height=700,
    margin=dict(l=40,r=40,t=40,b=40),
    paper_bgcolor="#111827",
    plot_bgcolor="#1F2937",

    font=dict(
        family="Arial",
        size=13,
        color="#F8FAFC"
    )
)

fig.update_traces(
    mode="lines+markers",

    line=dict(
        color="#2563EB",
        width=3,
        shape="spline",
        dash="solid"
    ),

    marker=dict(
        size=8,
        color="#2563EB",
        line=dict(
            color="white",
            width=2
        )
    ),

    opacity=0.95,

    hovertemplate=
        "<b>%{x}</b><br>"
        "Revenue: %{y}<extra></extra>"
)


fig.update_xaxes(
    title="Month",
    showgrid=False,
    showline=True,
    zeroline=True,
    tickfont=dict(
        size=14
    )

    
)

fig.update_yaxes(
    title="Revenue",
    showgrid=False,
    showline=True,
    zeroline=True,

    tickfont=dict(
        size=14
    ),
    title_font=dict(
        size=15
    )
)

fig.show()
```

```python
df3 = pd.DataFrame({
    "Quarter": ["Q1", "Q2", "Q3", "Q4"],
    "Profit_Margin": [0.18, 0.24, 0.21, 0.29]
})

fig = go.Figure()

fig.add_trace(
    go.Pie(
        labels=df3["Quarter"],
        values=df3["Profit_Margin"]
    )
)

fig.update_layout(
    title="Profit Distributions",
    width=600,
    height=600,
    margin=dict(l=40,r=40,t=40,b=40),
    paper_bgcolor="#111827",
    plot_bgcolor="#1F2937",

    font=dict(
        family="Arial",
        size=13,
        color="#F8FAFC"
    )
    
)


fig.update_traces(
    textinfo="label+percent",
    textposition="outside",

    marker=dict(
        colors=[
            "#2563EB",
            "#10B981",
            "#F59E0B",
            "#EF4444",
            "#8B5CF6"
        ],
        line=dict(
            color="#111827",
            width=2
        )
    ),

    pull=(0.02,0,0),

    opacity=0.95,

    hovertemplate=
        "<b>%{label}</b><br>"
        "Revenue: ₹%{value:,}<br>"
        "Share: %{percent}<extra></extra>"
)

fig.add_annotation(
    text="Q4 has the highest Profit Margin",
    x=0.5,
    y=1.15,
    showarrow=True,
    font=dict(
        size=14
    )
)

fig.show()
```
