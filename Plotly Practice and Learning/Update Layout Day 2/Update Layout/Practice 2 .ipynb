```python
df1 = pd.DataFrame({
    "Branch": ["Pune", "Mumbai", "Nashik", "Nagpur", "Kolhapur"],
    "Sales": [240000, 310000, 180000, 220000, 150000]
})

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df1["Branch"],
        y=df1["Sales"]
    )
)

fig.update_layout(
    title="Banch and Their Sales",
    width=800,
    height=700,
    xaxis_title="Branch",
    yaxis_title="Sales",
    margin=dict(
        l=50,
        r=60,
        t=50,
        b=50
    )
)

fig.show()
```

```python
df2 = pd.DataFrame({
    "Customer_Type": ["Regular", "Premium", "New", "Regular", "Premium"],
    "Orders": [45, 32, 18, 52, 40]
})

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=df2["Customer_Type"],
        y=df2["Orders"]
    )
)

fig.update_layout(
    title="Customer And Their Orders",
    width=800,
    height=500,
    xaxis_title="Customer Type",
    yaxis_title="Orders",
    margin=dict(
        l=50,
        r=50,
        t=50,
        b=50
    )
)

fig.show()
```

```python
df3 = pd.DataFrame({
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri"],
    "Visitors": [420, 580, 510, 690, 750]
})

fig = go.Figure()

fig.add_trace(
    go.Funnel(
        y=df3["Day"],
        x=df3["Visitors"]
    )
)

fig.update_layout(
    title="Days And Visitors",
    width=700,
    height=750,
    yaxis_title="Days",
    xaxis_title="Visitors",
    margin=dict(
        l=60,
        r=40,
        t=70,
        b=50
         
    )
)

fig.show()
```
