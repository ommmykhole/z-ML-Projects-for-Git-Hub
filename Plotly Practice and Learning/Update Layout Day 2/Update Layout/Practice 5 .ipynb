```python
df3 = pd.DataFrame({
    "Region": ["North", "South", "East", "West"],
    "Online": [120, 150, 100, 180],
    "Offline": [90, 110, 130, 140]
})

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df3["Region"],
        y=df3["Online"],
        mode="lines",
        fill="tozeroy",
        name="Online"
    )
)

fig.add_trace(
    go.Scatter(
        x=df3["Region"],
        y=df3["Offline"],
        mode="lines",
        fill="tozeroy",
        name="Offline"
    )
)

fig.update_layout(
    title="Sales By Region",
    width=800,
    height=600,
    xaxis_title="Region",
    yaxis_title="Online and Offline",

    margin=dict(l=40,r=40,t=40,b=40),
     template="plotly_dark",
    font=dict(
        size=15
    ),

    hovermode="y unified",

    plot_bgcolor="#0F172A",
    paper_bgcolor="#020617",

    showlegend=True,
    legend_title_text="Month",

    legend=dict(
        x=1,
        y=1
    ),

    xaxis=dict(
        showgrid=True,
        showline=True
    ),

    yaxis=dict(
        showgrid=True,
        showline=True
    )

    
)
  


fig.show()
```

```python
df1 = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Tablet", "Monitor", "Keyboard"],
    "Sales": [180, 240, 120, 95, 60]
})

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=df1["Product"],
        y=df1["Sales"],
        name="Sales"
    )
)

fig.update_layout(
    title="Product Sales",
    width=700,
    height=700,

    xaxis_title="Products",
    yaxis_title="Sales",

    margin=dict(
        l=40,
        r=30,
        t=60,
        b=60
    ),

    template="plotly_dark",

    font=dict(
        size=15
    ),

    plot_bgcolor="#020617",
    paper_bgcolor="#020617",

    showlegend=True,

    legend=dict(
        title_text="Product",
        x=1,
        y=1
    ),

    xaxis=dict(
        showgrid=True,
        showline=True,
        tickangle=-40,
        tickfont=dict(
            size=14
        )
    ),

    yaxis=dict(
        showgrid=True,
        showline=True,
        range=[0, 300]
    ),

    title_x=0.5,

    title_font=dict(
        size=15
    )
)

fig.show()
```
