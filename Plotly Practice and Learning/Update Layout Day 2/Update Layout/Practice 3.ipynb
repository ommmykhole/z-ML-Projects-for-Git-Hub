```python
df1 = pd.DataFrame({
    "Branch": ["Pune", "Mumbai", "Nashik", "Nagpur", "Kolhapur"],
    "Sales": [240000, 310000, 180000, 220000, 150000]
})

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=df1["Branch"],
        y=df1["Sales"],
    )
    
)

fig.update_layout(
    title="Branch and Their sales",
    width=500,
    height=500,
    xaxis_title="Branch",
    yaxis_title="Sales",
    margin=dict(l=50,r=60,t=60,b=70),
    template="plotly_dark",
    font=dict(size=15),
    plot_bgcolor="#111827",
    paper_bgcolor="#0B1120",
    hovermode="x unified"

)

fig.show()
```

```python
df2 = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Revenue": [120000, 145000, 135000, 175000, 190000, 220000]
})

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df2["Month"],
        y=df2["Revenue"]
    )
)

fig.update_layout(
    title="Month and Revenue",
    width=700,
    height=750,
    xaxis_title="Month",
    yaxis_title="Revenue",
    margin=dict(l=50,r=50,t=50,b=50),

    template="plotly_dark",
    font=dict(size=13),
    plot_bgcolor="#0F172A",
    paper_bgcolor="#020617",
    hovermode="y unified"
    
)

fig.show()
```

```python
df3 = pd.DataFrame({
    "Segment": ["Regular", "Premium", "New", "VIP"],
    "Customers": [420, 180, 250, 75]
})

fig = go.Figure()

fig.add_trace(
    go.Pie(
        labels=df3["Segment"],
        values=df3["Customers"]
    )
)

fig.update_layout(
    title="Customer Segmentation",
    width=500,
    height=500,
    xaxis_title="Segments",
    yaxis_title="Customer's",
    margin=dict(l=40,r=40,t=40,b=40),

    template="plotly_dark",

    plot_bgcolor="#1E1E1E",
    paper_bgcolor="#111111",

    font=dict(
        color="#E5E7EB",
        size=14
    ),

    legend=dict(
        font=dict(color="#D1D5DB")
    )

    
)

fig.show()
```
