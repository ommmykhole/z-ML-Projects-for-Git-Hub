```python
df1 = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Tablet", "Monitor", "Keyboard"],
    "Sales": [85, 120, 65, 45, 30]
})

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df1["Product"],
        y=df1["Sales"],
        fill="tozeroy"
    )
)

fig.update_layout(
    title="Product And Sales",
    width=600,
    height=600,
    xaxis_title="Product's",
    yaxis_title="Sales",
    margin=dict(l=40,r=40,t=50,b=40),

    template="plotly_dark",
    font=dict(
        size=14
        
    ),
    
    plot_bgcolor="#111111",
    paper_bgcolor="#1E1E1E",
    hovermode="x unified",

    showlegend=True,
    legend_title_text="Product",

    legend=dict(
        x=0,
        y=1
    ),
    xaxis=dict(
        showgrid=True,
        showline=True
    ),

    yaxis=dict(
        showgrid=True,
        showline=True
    )

    
)

fig.show()
```

```python
df2 = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Orders": [120, 150, 135, 180, 210, 240]
})

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df2["Month"],
        y=df2["Orders"],
        
    )
)

fig.update_layout(
    title="Month Wise Orders",
    width=700,
    height=740,
    xaxis_title="Month's",
    yaxis_title="Orders",
    margin=dict(l=40,r=40,t=40,b=40),

    template="plotly_dark",
    font=dict(
        size=15
    ),

    hovermode="y unified",

    plot_bgcolor="#0F172A",
    paper_bgcolor="#020617",

    showlegend=True,
    legend_title_text="Month",

    legend=dict(
        x=1,
        y=1
    ),

    xaxis=dict(
        showgrid=True,
        showline=True
    ),

    yaxis=dict(
        showgrid=True,
        showline=True
    )
)

fig.show()
```

```python
df3 = pd.DataFrame({
    "Region": ["North", "South", "East", "West"],
    "Online": [120, 150, 100, 180],
    "Offline": [90, 110, 130, 140]
})

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df3["Region"],
        y=df3["Online"],
        mode="lines",
        fill="tozeroy",
        name="Online"
    )
)

fig.add_trace(
    go.Scatter(
        x=df3["Region"],
        y=df3["Offline"],
        mode="lines",
        fill="tozeroy",
        name="Offline"
    )
)

fig.update_layout(
    title="Sales By Region",
    width=800,
    height=600,
    xaxis_title="Region",
    yaxis_title="Online and Offline",

    margin=dict(l=40,r=40,t=40,b=40),
     template="plotly_dark",
    font=dict(
        size=15
    ),

    hovermode="y unified",

    plot_bgcolor="#0F172A",
    paper_bgcolor="#020617",

    showlegend=True,
    legend_title_text="Month",

    legend=dict(
        x=1,
        y=1
    ),

    xaxis=dict(
        showgrid=True,
        showline=True
    ),

    yaxis=dict(
        showgrid=True,
        showline=True
    )

    
)
  


fig.show()
```
