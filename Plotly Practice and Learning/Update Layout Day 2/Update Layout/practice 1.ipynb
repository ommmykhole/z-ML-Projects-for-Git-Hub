```python
df1 = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [120, 180, 150, 220, 250]
})

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=df1["Month"],
        y=df1["Sales"]
    )
)

fig.update_layout(
    title="Month Wise Sales",
    width=500,
    height=500,
    xaxis_title="Month",
    yaxis_title="Sales"
)

fig.show()
```

```python
df2 = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Tablet", "Monitor", "Keyboard"],
    "Revenue": [85000, 65000, 42000, 30000, 12000]
})

fig = go.Figure()

fig.add_trace(
    go.Bar(
        y=df2["Product"],
        x=df2["Revenue"],
        orientation="h"
    )
)

fig.update_layout(
    title="Revenue On Product",
    width=700,
    height=790,
    yaxis_title="Product's",
    xaxis_title="Revenue",
    
)

fig.update_traces(
    marker=dict(
        opacity=0.8
    ),
    hovertemplate=
        "<b>%{y}</b><br>" +
        "Revenue: ₹%{x:,}<extra></extra>"
)

fig.show()
```

```python
df3 = pd.DataFrame({
    "Department": ["IT", "HR", "Sales", "Finance", "Marketing"],
    "Employees": [25, 10, 30, 15, 20]
})

fig = go.Figure()

fig.add_trace(
    go.Pie(
        labels=df3["Department"],
        values=df3["Employees"]
    )
)

fig.update_layout(
    title="Department And Employees",
    width=500,
    height=600,
    xaxis_title="Distributions"
)

fig.show()
```
