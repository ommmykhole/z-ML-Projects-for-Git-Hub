```python
import pandas as pd
import plotly.graph_objects as go
```

```python
df1 = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Tablet", "Monitor"],
    "Sales": [250, 180, 120, 150]
})


fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=df1["Product"],
        y=df1["Sales"]
    )
)

fig.update_layout(
    title="Product Sales",
    width=780,
    height=700,
    xaxis_title="Product",
    yaxis_title="Sales",
    margin=dict(
        l=30,r=40,t=40,b=30
    ),
    template="plotly_dark",
    plot_bgcolor="#020617",
    paper_bgcolor="#020617",

    showlegend=True
)

fig.update_traces(
    opacity=0.7,
    marker=dict(
        line=dict(
            width=2
        )
    ),

    text=df1["Sales"],

    textposition="inside",
    hovertemplate="product: %{x}<br>Sales: %{y}<extra></extra>"
)

fig.show()
```

```python
df2 = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Revenue": [120, 150, 135, 180, 210, 195]
})

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df2["Month"],
        y=df2["Revenue"],
        mode="lines+markers+text",
        name="Revenue"
    )
)

fig.update_layout(
    title="Month Wise Sales",
    width=700,
    height=700,

    xaxis_title="Month",
    yaxis_title="Revenue",

    margin=dict(
        l=40,
        r=40,
        t=60,
        b=40
    ),

    template="plotly_dark",
    plot_bgcolor="#020617",
    paper_bgcolor="#020617",
    title_x=0.5
)

fig.update_traces(
    opacity=0.8,

    marker=dict(
        size=8
    ),

    text=df2["Revenue"],
    textposition="top center"
)

fig.show()
```

```python
df3 = pd.DataFrame({
    "Age": [22, 25, 28, 32, 35, 40, 45, 50],
    "Spending": [120, 150, 180, 210, 260, 300, 340, 380]
})

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df3["Age"],
        y=df3["Spending"]
    )
)

fig.update_layout(
    title="Age And Spending",
    width=600,
    height=560,
    xaxis_title="Age",
    yaxis_title="Spending",
    margin=dict(l=40,r=30,t=40,b=30),
    template="plotly_white",
    plot_bgcolor="#020617",
    paper_bgcolor="#020617",
    title_x=0.5,
    xaxis=dict(showgrid=False),
    yaxis=dict(showgrid=False)
)

fig.update_traces(
    opacity=0.7,
    marker=dict(
        size=6
    ),
    mode="markers",
    textposition="top center",
    hovertemplate="Age: %{x}<br>Spending: %{y}<extra></extra>"
)

fig.show()
```

```python
df4 = pd.DataFrame({
    "Sales": [120, 150, 180, 200, 210, 220, 250, 270,
              280, 300, 310, 330, 350, 380, 400]
})

fig = go.Figure()

fig.add_trace(
    go.Histogram(
        x=df4["Sales"],
        name="Sales"
    )
)

fig.update_layout(
    title="Sales",
    width=700,
    height=670,
    margin=dict(
        l=40,
        r=40,
        t=30,
        b=30
    ),
    template="plotly_dark",
    plot_bgcolor="#020617",
    paper_bgcolor="#020617",
    title_x=0.5,
    xaxis=dict(showgrid=False),
    yaxis=dict(showgrid=False)
    
)

fig.update_traces(
    opacity=0.7,
    text=df4["Sales"],
    textposition="outside"
    
)

fig.show()
```

```python
df5 = pd.DataFrame({
    "Category": ["Electronics", "Clothing", "Grocery", "Furniture"],
    "Revenue": [420, 280, 350, 190]
})

fig = go.Figure()

fig.add_trace(
    go.Pie(
        labels=df5["Category"],
        values=df5["Revenue"],
        hole=0.5
    )
)

fig.update_layout(
    title="Category Wise Revenue",
    width=700,
    height=600,
    template="plotly_dark",
    margin=dict(l=40,r=40,t=40,b=40),
    title_x=0.5
)

fig.update_traces(
    opacity=0.6,
    text=df5["Category"],
    textposition="outside"
)

fig.show()
```
