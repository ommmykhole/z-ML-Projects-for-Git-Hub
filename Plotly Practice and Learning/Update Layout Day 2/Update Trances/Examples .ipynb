```python
import pandas as pd
import plotly.graph_objects as go

df = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Tablet"],
    "Sales": [250, 180, 120]
})

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=df["Product"],
        y=df["Sales"],
        name="Sales"
    )
)

fig.update_traces(
    
    # 1. Transparency
    opacity=0.9,

    # 2. Marker / bar appearance
    marker=dict(
        size=15
    ),

    # 3. Line appearance
    # line=dict(
    #     width=2
    # ),

    # 4. Text on bars
    text=df["Sales"],

    # 5. Text position
    textposition="outside",

    # 6. Hover customization
    hovertemplate="Product: %{x}<br>Sales: %{y}<extra></extra>",

    # 7. Hover information
    # hoverinfo="x+y",

    # 8. Mode
    # mode="lines+markers",

)

fig.show()
```

```python
import pandas as pd
import plotly.graph_objects as go

df = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Tablet"],
    "Sales": [250, 180, 120]
})

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=df["Product"],
        y=df["Sales"],
        name="Sales"
    )
)

fig.update_layout(
    title="Product Sales",
    width=600,
    height=600,
    xaxis_title="Product",
    yaxis_title="Sales",
    margin=dict(
        l=40,
        r=30,
        t=60,
        b=60
    ),

    template="plotly_dark",

    font=dict(size=15),

    plot_bgcolor="#020617",
    paper_bgcolor="#020617",

    showlegend=True,
)

fig.update_traces(
    # 1. Transparency
    opacity=0.5,

    # 2. Bar appearance
    marker=dict(
        line=dict(
            width=2
        )
    ),

    # 3. Text on bars
    text=df["Sales"],

    # 4. Text position
    textposition="outside",

    # 5. Hover customization
    hovertemplate="Product: %{x}<br>Sales: %{y}<extra></extra>"
)

fig.show()
```

```python
import pandas as pd
import plotly.graph_objects as go

df = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Tablet", "Monitor"],
    "Sales": [250, 180, 120, 150]
})

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=df["Product"],
        y=df["Sales"],
        name="Sales"
    )
)

fig.update_layout(
    title="Product Wise Sales",
    width=800,
    height=790,
    xaxis_title="Product",
    yaxis_title="Sales",
    margin=dict(
        l=50,
        r=50,
        t=40,
        b=40
    ),
    template="plotly_dark",
    plot_bgcolor="#020617",
    paper_bgcolor="#020617",
)

fig.update_traces(
    opacity=0.5,

    text=df["Sales"],
    textposition="outside",

    hovertemplate=
        "Product: %{x}<br>"
        "Sales: %{y}<extra></extra>"
)

fig.show()
```
