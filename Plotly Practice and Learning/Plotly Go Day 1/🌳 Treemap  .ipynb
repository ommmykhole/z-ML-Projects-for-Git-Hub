```python
import pandas as pd
import plotly.graph_objects as go

df1 = pd.DataFrame({
    "Category": ["Electronics", "Electronics", "Electronics",
                 "Clothing", "Clothing", "Clothing"],
    "Product": ["Laptop", "Mobile", "Tablet",
                "Shirt", "Jeans", "Jacket"],
    "Sales": [120000, 95000, 65000, 50000, 75000, 40000]
})

# Category totals
category_sales = df1.groupby("Category")["Sales"].sum()

# Labels
labels = list(category_sales.index) + list(df1["Product"])

# Parents
parents = [""] * len(category_sales) + list(df1["Category"])

# Values
values = list(category_sales.values) + list(df1["Sales"])

fig = go.Figure()

fig.add_trace(
    go.Treemap(
        labels=labels,
        parents=parents,
        values=values
    )
)

fig.show()
```
