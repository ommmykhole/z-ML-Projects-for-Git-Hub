```python
import plotly.graph_objects as go
import pandas as pd

df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],

    "Sales": [45000, 52000, 48000, 61000, 68000, 72000,
              65000, 78000, 85000, 81000, 92000, 105000]
})

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df["Month"],
        y=df["Sales"],
        mode="lines"
    )
)

fig.show()
```
