```python
import plotly.graph_objects as go
import pandas as pd
```

```python
fig = go.Figure()

fig.add_trace(
    go.Funnel(
        y=["Visitors", "Leads", "Enquiries", "Orders", "Customers"],
        x=[5000, 2500, 1200, 650, 300]
    )
)

fig.show()
```

```python
df1 = pd.DataFrame({
    "Stage": ["Visitors", "Product Views", "Add to Cart", "Checkout", "Orders"],
    "Count": [5000, 3200, 1800, 950, 520]
})

fig = go.Figure()

fig.add_trace(
    go.Funnel(
        y=df1["Stage"],
        x=df1["Count"],
        name="Stage"
    )
)
   

fig.show()
```
