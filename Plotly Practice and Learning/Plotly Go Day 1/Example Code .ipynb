```python
import plotly.graph_objects as go

# ---------------- DATA ----------------
products = ["Shirt", "Jeans", "Shoes", "Jacket"]

sales = [50000, 75000, 60000, 40000]

# ---------------- FIGURE ----------------
fig = go.Figure()

# ---------------- BAR TRACE ----------------
fig.add_trace(
    go.Bar(
        x=products,
        y=sales,
        name="Sales",
        text=sales,
        textposition="outside",
        hovertemplate=
            "<b>%{x}</b><br>" +
            "Sales: ₹%{y:,}<extra></extra>"
    )
)

# ---------------- SLICER / DROPDOWN ----------------
fig.update_layout(
    updatemenus=[
        dict(
            buttons=[
                dict(
                    label="All Products",
                    method="update",
                    args=[
                        {"visible": [True]},
                        {"title": "Sales - All Products"}
                    ]
                ),

                dict(
                    label="Top Products",
                    method="update",
                    args=[
                        {"visible": [True]},
                        {"title": "Sales - Top Products"}
                    ]
                )
            ],
            direction="down",
            showactive=True,
            x=0.85,
            y=1.15
        )
    ],

    # ---------------- CUSTOMIZATION ----------------
    title="Product Sales Dashboard",

    xaxis=dict(
        title="Product",
        showgrid=False
    ),

    yaxis=dict(
        title="Sales (₹)",
        tickprefix="₹",
        gridcolor="lightgray"
    ),

    template="plotly_white",

    height=500,

    margin=dict(
        l=70,
        r=40,
        t=100,
        b=70
    )
)

# ---------------- SHOW ----------------
fig.show()
```

```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Shirt", "Jeans", "Shoes", "Jacket", "Cap"],
    "Sales": [45000, 72000, 58000, 39000, 22000]
})

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=df["Product"],
        y=df["Sales"]
    )
)

fig.show()
```
