```python
df1 = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [45000, 52000, 48000, 61000, 68000, 75000],
    "Profit": [12000, 15000, 13000, 18000, 21000, 25000],
    "Expenses": [33000, 37000, 35000, 43000, 47000, 50000]
})

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df1["Month"],
        y=df1[["Sales","Profit","Expenses"]],
        fill="tozeroy"
    )
)

fig.show()
```

```python
fig = go.Figure()

# Sales Trace
fig.add_trace(
    go.Scatter(
        x=df1["Month"],
        y=df1["Sales"],
        mode="lines",
        fill="tozeroy",
        name="Sales"
    )
)

# Profit Trace
fig.add_trace(
    go.Scatter(
        x=df1["Month"],
        y=df1["Profit"],
        mode="lines",
        fill="tozeroy",
        name="Profit"
    )
)

# Expenses Trace
fig.add_trace(
    go.Scatter(
        x=df1["Month"],
        y=df1["Expenses"],
        mode="lines",
        fill="tozeroy",
        name="Expenses"
    )
)

fig.show()
```
