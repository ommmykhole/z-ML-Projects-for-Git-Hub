```python
st.set_page_config(
    page_title="EV Charging Inteligence",
    page_icon="⚡",
    layout="wide"
)


st.sidebar.header("Globel Filter")
st.sidebar.caption("Apply Filters From Given Filters")


# Date Range
date_range = st.sidebar.date_input(
    "Date Range",
    value=(
        df["StartDatetime"].min().date(),
        df["StartDatetime"].max().date()
    )
)

# Location
location = st.sidebar.multiselect(
    "Location",
    options=sorted(df["Location"].dropna().unique()),
    default=sorted(df["Location"].dropna().unique())
)


# Charger Type
charger_type = st.sidebar.multiselect(
    "Charger Type",
    options=sorted(df["ChargerType"].dropna().unique()),
    default=sorted(df["ChargerType"].dropna().unique())
)



# AC / DC
ac_dc = st.sidebar.multiselect(
    "Charging Type",
    options=sorted(df["ChargerACDC"].dropna().unique()),
    default=sorted(df["ChargerACDC"].dropna().unique())
)



# Charger Capacity
capacity = st.sidebar.multiselect(
    "Charger Capacity",
    options=sorted(df["ChargerCapacity"].dropna().unique()),
    default=sorted(df["ChargerCapacity"].dropna().unique())
)



# Charger Company
company = st.sidebar.multiselect(
    "Charger Company",
    options=sorted(df["ChargerCompany"].dropna().unique()),
    default=sorted(df["ChargerCompany"].dropna().unique())
)









# DF Copy
filtered_df = df.copy()



# Date Filter
start_date = pd.to_datetime(date_range[0])
end_date = pd.to_datetime(date_range[1]) + pd.Timedelta(days=1)

filtered_df = filtered_df[
    (filtered_df["StartDatetime"] >= start_date) &
    (filtered_df["StartDatetime"] < end_date)
]



# Location Filter
filtered_df = filtered_df[
    filtered_df["Location"].isin(location)
]



# Charger Filter
filtered_df = filtered_df[
    filtered_df["ChargerType"].isin(charger_type)
]



# AC/DC Filter 
filtered_df = filtered_df[
    filtered_df["ChargerACDC"].isin(ac_dc)
]


#  Charger Capacity
filtered_df = filtered_df[
    filtered_df["ChargerCapacity"].isin(capacity)
]


# Company 
filtered_df = filtered_df[
    filtered_df["ChargerCompany"].isin(company)
]

```

```python
@st.cache_data
def load_data():
    df = pd.read_csv("data/EV_Charging.csv")

    df["StartDatetime"] = pd.to_datetime(df["StartDatetime"])
    df["EndDatetime"] = pd.to_datetime(df["EndDatetime"])

    return df


df = load_data()
```
